# Análisis Alura Store

## Paso 1: Exploración de Datos

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
store1 = pd.read_csv('../datos/tienda_1.csv')
store2 = pd.read_csv('../datos/tienda_2.csv')
store3 = pd.read_csv('../datos/tienda_3.csv')
store4 = pd.read_csv('../datos/tienda_4.csv')
```

```python
store1.head()
```

```python
print('Columnas:')
print(store1.columns.tolist())
```

```python
print('Registros por tienda:')
print(f'Store 1: {len(store1)}')
print(f'Store 2: {len(store2)}')
print(f'Store 3: {len(store3)}')
print(f'Store 4: {len(store4)}')
```

```python
print('Valores faltantes:')
print(store1.isnull().sum())
```

### Ingresos Totales

```python
ingresos_1 = store1['Precio'].sum()
ingresos_2 = store2['Precio'].sum()
ingresos_3 = store3['Precio'].sum()
ingresos_4 = store4['Precio'].sum()

print('Ingresos totales:')
print(f'Store 1: ${ingresos_1:,.2f}')
print(f'Store 2: ${ingresos_2:,.2f}')
print(f'Store 3: ${ingresos_3:,.2f}')
print(f'Store 4: ${ingresos_4:,.2f}')
print(f'Total: ${ingresos_1 + ingresos_2 + ingresos_3 + ingresos_4:,.2f}')
```

```python
print('Estadísticas por tienda:\n')
stores = [store1, store2, store3, store4]
for i, store in enumerate(stores, 1):
    print(f'Store {i}:')
    print(f'  Transacciones: {len(store)}')
    print(f'  Ingreso total: ${store["Precio"].sum():,.2f}')
    print(f'  Precio promedio: ${store["Precio"].mean():,.2f}')
    print(f'  Mínimo: ${store["Precio"].min():,.2f}')
    print(f'  Máximo: ${store["Precio"].max():,.2f}')
    print()
```

### Gráfico 1: Ingresos Totales

```python
tiendas_nombres = ['Store 1', 'Store 2', 'Store 3', 'Store 4']
ingresos = [ingresos_1, ingresos_2, ingresos_3, ingresos_4]

plt.figure(figsize=(10, 6))
plt.bar(tiendas_nombres, ingresos, color=['blue', 'green', 'orange', 'red'])
plt.xlabel('Tienda')
plt.ylabel('Ingresos ($)')
plt.title('Ingresos Totales por Tienda')
for i, v in enumerate(ingresos):
    plt.text(i, v, f'${v/1e9:.2f}B', ha='center', va='bottom')
plt.tight_layout()
plt.savefig('../resultados/grafico_ingresos.png', dpi=300)
plt.show()
```

## Paso 2: Productos por Categoría

```python
print('Productos vendidos por categoría:\n')
for i, store in enumerate(stores, 1):
    print(f'Store {i}:')
    categorias = store['Categoría del Producto'].value_counts()
    print(categorias)
    print()
```

### Gráfico 2: Categorías

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Distribución de Categorías por Tienda', fontsize=16)

for i, store in enumerate(stores):
    ax = axes[i // 2, i % 2]
    categorias = store['Categoría del Producto'].value_counts()
    ax.pie(categorias.values, labels=categorias.index, autopct='%1.1f%%')
    ax.set_title(f'Store {i+1}')

plt.tight_layout()
plt.savefig('../resultados/grafico_categorias.png', dpi=300)
plt.show()
```

## Paso 3 y 4: Productos Más y Menos Vendidos

```python
print('Productos más vendidos:\n')
for i, store in enumerate(stores, 1):
    print(f'Store {i}:')
    productos = store['Producto'].value_counts().head(5)
    print(productos)
    print()
```

```python
print('Productos menos vendidos:\n')
for i, store in enumerate(stores, 1):
    print(f'Store {i}:')
    productos = store['Producto'].value_counts().tail(5)
    print(productos)
    print()
```

### Gráfico 3: Top 10 Productos - Store 1

```python
top_productos = store1['Producto'].value_counts().head(10)

plt.figure(figsize=(12, 6))
plt.barh(range(len(top_productos)), top_productos.values, color='steelblue')
plt.yticks(range(len(top_productos)), top_productos.index)
plt.xlabel('Cantidad Vendida')
plt.title('Top 10 Productos Más Vendidos - Store 1')
plt.tight_layout()
plt.savefig('../resultados/grafico_productos_top.png', dpi=300)
plt.show()
```

## Paso 5: Costo de Envío Promedio

```python
print('Costo de envío promedio por tienda:\n')
for i, store in enumerate(stores, 1):
    costo_promedio = store['Costo de envío'].mean()
    costo_total = store['Costo de envío'].sum()
    print(f'Store {i}:')
    print(f'  Costo promedio: ${costo_promedio:,.2f}')
    print(f'  Costo total: ${costo_total:,.2f}')
    print()
```

### Gráfico 4: Costo de Envío

```python
costos_promedio = [store1['Costo de envío'].mean(),
                   store2['Costo de envío'].mean(),
                   store3['Costo de envío'].mean(),
                   store4['Costo de envío'].mean()]

plt.figure(figsize=(10, 6))
plt.bar(tiendas_nombres, costos_promedio, color=['blue', 'green', 'orange', 'red'])
plt.xlabel('Tienda')
plt.ylabel('Costo Promedio ($)')
plt.title('Costo de Envío Promedio por Tienda')
for i, v in enumerate(costos_promedio):
    plt.text(i, v, f'${v:,.0f}', ha='center', va='bottom')
plt.tight_layout()
plt.savefig('../resultados/grafico_envios.png', dpi=300)
plt.show()
```

## Resumen Comparativo

```python
resumen = []
for i, store in enumerate(stores, 1):
    resumen.append({
        'Tienda': f'Store {i}',
        'Ingresos': store['Precio'].sum(),
        'Transacciones': len(store),
        'Precio Promedio': store['Precio'].mean(),
        'Rating Promedio': store['Calificación'].astype(int).mean(),
        'Costo Envío Promedio': store['Costo de envío'].mean()
    })

df_resumen = pd.DataFrame(resumen)
print(df_resumen.to_string())
```

## Informe Final

### Introducción

Este análisis fue realizado para ayudar al Sr. Juan a identificar cuál de sus cuatro tiendas de Alura Store debería vender para iniciar un nuevo emprendimiento. Se evaluaron múltiples factores incluyendo ingresos, rendimiento operacional, satisfacción del cliente y eficiencia logística.

### Resultados Clave

**1. Ingresos Totales:**
- Store 1: $1,150,880,400.00 (Más alto)
- Store 2: $1,116,343,500.00
- Store 3: $1,098,019,600.00
- Store 4: $1,038,375,700.00 (Más bajo)

**2. Rating Promedio de Clientes:**

```python
ratings = []
for i, store in enumerate(stores, 1):
    rating = store['Calificación'].astype(int).mean()
    ratings.append(rating)
    print(f'Store {i}: {rating:.2f}/5')
```

**3. Análisis de Fortalezas y Debilidades:**

**Store 1:**
- Ingresos más altos: $1,150,880,400
- Precio promedio más elevado
- Tienda más eficiente y competitiva

**Store 2:**
- Ingresos moderados: $1,116,343,500
- Segundo lugar en performance
- Operación estable

**Store 3:**
- Ingresos en tercer lugar: $1,098,019,600
- Performance media
- Oportunidades de mejora

**Store 4:**
- Ingresos más bajos: $1,038,375,700 (11% menos que Store 1)
- Rating promedio más bajo
- Precio promedio menor
- Menos transacciones con bajo retorno
- Costo de envío promedio mayor sin compensar en volumen

### Conclusión y Recomendación

```python
print('='*60)
print('RECOMENDACIÓN FINAL')
print('='*60)
print('\nSe recomienda vender: STORE 4')
print('\nJustificación:')
print('\n1. Ingresos: Store 4 genera los ingresos más bajos')
print('   ($1,038,375,700), representando el 90% del ingreso')
print('   de Store 1. Diferencia de $112.5 millones.')
print('\n2. Satisfacción del Cliente: Rating promedio más bajo,')
print('   indicando menor satisfacción de los clientes.')
print('\n3. Eficiencia: Precio promedio menor sugiere menor')
print('   capacidad de vender productos de mayor valor.')
print('\n4. Logística: Costo de envío promedio más alto sin')
print('   proporcionar retorno proporcional en ingresos.')
print('\n5. Margen Operacional: La combinación de menores')
print('   ingresos, baja satisfacción y altos costos de envío')
print('   indica márgenes operacionales reducidos.')
print('\nConclusion: Store 4 es la menos eficiente del grupo')
print('y es el candidato más apropiado para vender.')
print('='*60)
```

## Extra: Análisis Geográfico (Opcional)

### Distribución Geográfica de Ventas

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 12))
fig.suptitle('Distribución Geográfica de Ventas por Tienda', fontsize=16)

for i, store in enumerate(stores):
    ax = axes[i // 2, i % 2]
    scatter = ax.scatter(store['lon'], store['lat'], 
                         c=store['Precio'], cmap='viridis', 
                         alpha=0.6, s=30)
    ax.set_xlabel('Longitud')
    ax.set_ylabel('Latitud')
    ax.set_title(f'Store {i+1}')
    plt.colorbar(scatter, ax=ax, label='Precio ($)')

plt.tight_layout()
plt.savefig('../resultados/grafico_distribucion_geografica.png', dpi=300)
plt.show()
```

### Análisis de Ciudades/Ubicaciones

```python
print('Lugares de compra por tienda:\n')
for i, store in enumerate(stores, 1):
    print(f'Store {i}:')
    lugares = store['Lugar de Compra'].value_counts()
    print(lugares)
    print()
```

### Concentración de Ventas por Región

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Ingresos por Ubicación Geográfica', fontsize=16)

for i, store in enumerate(stores):
    ax = axes[i // 2, i % 2]
    lugar_ingresos = store.groupby('Lugar de Compra')['Precio'].sum().sort_values()
    lugar_ingresos.plot(kind='barh', ax=ax, color='steelblue')
    ax.set_xlabel('Ingresos ($)')
    ax.set_title(f'Store {i+1}')

plt.tight_layout()
plt.savefig('../resultados/grafico_ingresos_por_ubicacion.png', dpi=300)
plt.show()
```
